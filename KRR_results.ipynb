```python
import os
import pandas as pd
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
from IPython.display import display, Markdown

def highlight_min(row, column_index):
    styles = []
    for col in row.index:
        top_level = col[0]  
        # print(top_level)
        if top_level == 'MSE':
            max_value = row[column_index.get_loc(top_level)].min()
            second_value = row[column_index.get_loc(top_level)].nsmallest(2).iloc[-1]
            third_value = row[column_index.get_loc(top_level)].nsmallest(3).iloc[-1]
        elif top_level == 'R2':
            max_value = row[column_index.get_loc(top_level)].max()
            second_value = row[column_index.get_loc(top_level)].nlargest(2).iloc[-1]
            third_value = row[column_index.get_loc(top_level)].nlargest(3).iloc[-1]
        style = ''
        if row[col] == max_value:
            # style = 'background-color: #e3867f'
            style = 'background-color: #cfb04a'
        elif row[col] == second_value:
            style = 'background-color: #dbd6c5; color: #333333'
        elif row[col] == third_value:
            style = 'background-color: #785e08'
        styles.append(style)
    return styles

def show_regression_results(label: str, source_df: pd.DataFrame):
    print(f'{label}:')
    # column_replace = ('Metric', 'Model')
    # if column_replace in source_df.columns:
    #     source_df = source_df.drop(columns=column_replace, inplace=False)
    #     source_df.columns.names = list(('Metric', 'Model'))
    means = source_df.mean()
    means_df = means.to_frame().T
    styled_mean_df = means_df.style.apply(
        highlight_min, axis=1, column_index=means_df.columns
    )
    display(styled_mean_df)

def show_combined_regression_results(title: str, source_df: pd.DataFrame):
    display(Markdown(f'## {title}'))
    styled_df = source_df.style.apply(
        highlight_min, axis=1, column_index=source_df.columns
    )
    display(styled_df)

def show_regression_results(label: str, source_df: pd.DataFrame):
    print(f'{label}:')
    means = source_df.mean()
    means_df = means.to_frame().T
    styled_mean_df = means_df.style.apply(
        highlight_min, axis=1, column_index=means_df.columns
    )
    display(styled_mean_df)

def show_single_multilevel_table_results(title: str, show: list):
    rows = []
    dfs = []
    for show_title_parts, results_df in show:
        norm, weights, reg = show_title_parts
        w_pow, w_r_opt = weights.split(' ', 1)
        rows.append(tuple([norm, w_pow, w_r_opt, reg]))
        means = results_df.mean()
        means_df = means.to_frame().T
        dfs.append(means_df)
        # show_regression_results(show_title, results_df)
    row_index = pd.MultiIndex.from_tuples(rows, names=["Normalization", "Weights", "R Opt", "Regularization"])
    combined_df = pd.concat(dfs, ignore_index=False)
    # print(combined_df)
    combined_df.index = row_index
    sorted_df = combined_df.sort_index(level=["Normalization", "Weights", "R Opt", "Regularization"])
    show_combined_regression_results(title, sorted_df)
    return sorted_df
```
